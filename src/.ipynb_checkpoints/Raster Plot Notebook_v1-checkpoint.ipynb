```python
import pandas as pd
import matplotlib.pyplot as plt
import ast

# === 1. Build the dataframe ===
spike_df = load_spike_data()
df = match_times(spike_df, progress=True)

# === 2. Define the mega_raster() function ===
def mega_raster(df, window=0.1):
    """
    Generate a merged raster plot of cluster spiking aligned to ripple peaks.

    Parameters
    ----------
    df : pandas.DataFrame
        DataFrame with at least these columns:
        - "Peak"
        - "Spike Times (s)"
        - "Cluster IDs"
    window : float, optional
        Time window (in seconds) to show around ripple peak (default ±0.1 s).
    """

    # --- Convert stringified lists to real lists (if needed) ---
    for col in ["Spike Times (s)", "Cluster IDs"]:
        if isinstance(df[col].iloc[0], str):
            df[col] = df[col].apply(ast.literal_eval)

    # --- Expand each ripple into a long-form dataframe ---
    expanded_rows = []
    for _, row in df.iterrows():
        peak = float(row["Peak"])
        spike_times = [float(s) for s in row["Spike Times (s)"]]
        clusters = row["Cluster IDs"]

        for spk_time, clust in zip(spike_times, clusters):
            expanded_rows.append({
                "ripple_peak": peak,
                "spike_time": spk_time,
                "cluster_id": clust,
                "t_rel": spk_time - peak  # time relative to ripple peak
            })

    expanded_df = pd.DataFrame(expanded_rows)

    # --- Filter to ±window seconds around each ripple ---
    mask = (expanded_df["t_rel"] >= -window) & (expanded_df["t_rel"] <= window)
    plot_df = expanded_df.loc[mask]

    print(f"Plotting {len(plot_df)} spikes across {len(df)} ripples")

    # --- Raster plot ---
    fig, ax = plt.subplots(figsize=(9, 6))
    ax.scatter(plot_df["t_rel"], plot_df["cluster_id"], s=3, color="black", alpha=0.7)
    ax.axvline(0, color="red", linestyle="--", linewidth=1)

    # --- Formatting ---
    ax.set_xlim(-window, window)
    ax.set_xlabel("Time (s) relative to ripple peak")
    ax.set_ylabel("Cluster ID")
    ax.set_title(f"Mega raster aligned to ripple peaks (±{window*1000:.0f} ms)")

    plt.tight_layout()
    plt.show()

# === 3. Call the function ===
mega_raster(df, window=0.1)
```

```python
# add new columns to pandas df with:
# 1. neuron cluster sequence (omit repeats, preserve order based on first spike) 
# 2. cluster IDs ordered smallest to largest (still omtting repeats)

# --- make df if not already created with match_times ---
df = match_times(
     spike_df,
     swr_dir="./SWRs_7744_partner_intro.csv",
     only_keep_good=False,
     progress=False
)

# --- Function to remove duplicates in the cluster list of a SWR ---
def clusters_in_SWR(seq):
    seen = set()
    return [x for x in seq if not (x in seen or seen.add(x))]

# --- Add new columns to df
df["Unique_Cluster_IDs"] = df["Cluster IDs"].apply(clusters_in_SWR) 
df["Sorted_Cluster_IDs"] = df["Unique_Cluster_IDs"].apply(lambda lst: sorted(lst))

print(f"Added 'Unique_Cluster_IDs' and 'Sorted_Cluster_IDs' columns")
```

```python
# --- Export to CSV for human readable file ---
timestamp = datetime.now().strftime("%Y%m%d_%H%M")
df.to_csv(f"SWRsClusterID_{timestamp}.csv", index=False)

# --- View CSV ---
df = pd.read_csv("SWRsClusterID.csv")
# print(df.columns.tolist())
# print(df.head())

print(f"Saved to CSV")
```

```python
# --- save df as a pickle file for loading in raster plots ---
timestamp = datetime.now().strftime("%m%d_%H%M")
df.to_pickle(f"df_{timestamp}.pkl")

print(f"Saved pickle file as: df_{timestamp}.pkl")
```

MAKE RASTER PLOTS

```python
# Make one raster plot of cluster spiking that is the merge of all SWRs aligned to thier peak

# load data
#df = pd.read_pickle("df_1020_1358.pkl")

# === Expand each ripple into long-form dataframe ===
expanded_rows = []
for _, row in df.iterrows():
    peak = row["Peak"]
    spike_times = row["Spike Times (s)"]
    clusters = row["Cluster IDs"]
    
    # Ensure matching lengths
    for spk_time, clust in zip(spike_times, clusters):
        expanded_rows.append({
            "ripple_peak": peak,
            "spike_time": spk_time,
            "cluster_id": clust,
            "t_rel": spk_time - peak  # time relative to ripple peak
        })

expanded_df = pd.DataFrame(expanded_rows)

# === Filter to ±100 ms window (0.1 s) around each ripple ===
window = 0.1
mask = (expanded_df["t_rel"] >= -window) & (expanded_df["t_rel"] <= window)
plot_df = expanded_df.loc[mask]

print(f"Plotting {len(plot_df)} spikes across {len(df)} ripples")

# === Raster plot ===
fig, ax = plt.subplots(figsize=(9, 6))
ax.scatter(plot_df["t_rel"], plot_df["cluster_id"], s=3, color="black", alpha=0.7)
ax.axvline(0, color="red", linestyle="--", linewidth=1)

ax.set_xlim(-window, window)
ax.set_xlabel("Time (s) relative to ripple peak")
ax.set_ylabel("Cluster ID")
ax.set_title("Spike raster aligned to ripple peaks (±100 ms)")
plt.tight_layout()
plt.show()
```

```python
# plot one raster plot at a time:

# load data
#df = pd.read_pickle("df_1020_1358.pkl")

# === Choose which ripple to plot (0 = first row) ===
ripple_index = 10  # ← change this number to view a different ripple
ripple = df.iloc[ripple_index]

# Extract relevant info
peak_time = ripple["Peak"]
spike_times = ripple["Spike Times (s)"]
cluster_ids = ripple["Cluster IDs"]

# Check that both lists match
assert len(spike_times) == len(cluster_ids), "Spike Times and Cluster IDs length mismatch!"

# Compute time relative to ripple peak
t_rel = [t - peak_time for t in spike_times]

# === Define plotting window (±100 ms) ===
window = 0.05
x_min, x_max = -window, window

# === Raster plot for this ripple ===
fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(t_rel, cluster_ids, s=20, color="black", marker='|')
ax.axvline(0, color="red", linestyle="--", linewidth=1)
ax.set_xlim(x_min, x_max)
ax.set_xlabel("Time (s) relative to ripple peak")
ax.set_ylabel("Cluster ID")
ax.set_title(f"Ripple #{ripple_index} | Peak = {peak_time:.3f}s | {len(spike_times)} spikes")
plt.tight_layout()
plt.show()
```

```python
#scale y-axis for only the clusters in the SWR event

# === Choose which ripple to plot (0 = first row) ===
ripple_index = 130  # change this number to view a different ripple
ripple = df.iloc[ripple_index]

# Extract relevant info
peak_time = ripple["Peak"]
spike_times = ripple["Spike Times (s)"]
cluster_ids = ripple["Cluster IDs"]

# Check matching lengths
assert len(spike_times) == len(cluster_ids), "Spike Times and Cluster IDs length mismatch!"

# Compute time relative to ripple peak
t_rel = [t - peak_time for t in spike_times]

# === Define plotting window (±50 ms) ===
window = 0.05
x_min, x_max = -window, window

# === Scale y-axis spacing ===
scale_y = 2.0  # increase to make more vertical space between clusters
unique_clusters = sorted(set(cluster_ids))

# Map each cluster ID to a scaled position
cluster_to_scaled = {cid: i * scale_y for i, cid in enumerate(unique_clusters)}
scaled_y = [cluster_to_scaled[c] for c in cluster_ids]

# === Raster plot for this ripple ===
fig, ax = plt.subplots(figsize=(8, 6))  # still one plot, just with one axis
ax.scatter(t_rel, scaled_y, s=50, color="black", marker='|')

# Add reference line
ax.axvline(0, color="red", linestyle="--", linewidth=1)

# Update y-axis ticks to show cluster IDs
ax.set_yticks([cluster_to_scaled[c] for c in unique_clusters])
ax.set_yticklabels([str(int(c)) for c in unique_clusters])

# Axis formatting
ax.set_xlim(x_min, x_max)
ax.set_xlabel("Time (s) relative to ripple peak")
ax.set_ylabel("Active Cluster IDs")
ax.set_title(f"Ripple #{ripple_index} | Peak = {peak_time:.3f}s | {len(spike_times)} spikes")

plt.tight_layout()
plt.show()
```
